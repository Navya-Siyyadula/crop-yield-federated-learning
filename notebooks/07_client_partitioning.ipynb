```python
import os
import pandas as pd

# Load training data
X_train = pd.read_csv("/content/agri_dataset/X_train.csv")
y_train = pd.read_csv("/content/agri_dataset/y_train.csv")

# Combine X and y
data = pd.concat([X_train, y_train], axis=1)

# Shuffle
data = data.sample(frac=1, random_state=42).reset_index(drop=True)

# Create folder
output_folder = "/content/edge_clients"
os.makedirs(output_folder, exist_ok=True)

# Correct split: 88 + 88 + 88 + 86 = 350
split_sizes = [88, 88, 88, 86]

start = 0

for i, size in enumerate(split_sizes, start=1):
    client_data = data.iloc[start:start + size]

    X_client = client_data.iloc[:, :38]
    y_client = client_data.iloc[:, 38:]

    X_client.to_csv(f"{output_folder}/client_{i}_X.csv", index=False)
    y_client.to_csv(f"{output_folder}/client_{i}_y.csv", index=False)

    print(f"Client {i}: {size} samples")

    start += size

print("\nDataset successfully split into 4 edge clients!")
```
